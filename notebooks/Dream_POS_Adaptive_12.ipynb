# Dream-7B POS fixed-12 replica

Scientific replica of the DiffuLLaMA fixed-12 protocol: all-head `.50` screens at early/middle/late, 12 frozen candidates per depth, held-out confirmation, and selected-head `.25/.50/.75` trajectories. Dream results and checkpoints remain completely separate.

If Dream POS extraction is incomplete, run the explicitly gated extraction cell on a GPU first. Then change Colab to a 12-logical-CPU runtime and rerun with `RUN_EXTRACTION = False`.

```python
import json, os, shutil, subprocess
from pathlib import Path
from google.colab import drive

REPOSITORY = 'https://github.com/Dabsoysauce/latentrelationsondlm.git'
BRANCH = 'optimize/adaptive-pos-cpu-under-3h'
PROTOCOL_COMMIT = '3da9dede1b43308c0583083c9ec70b6139134d01'
RUN_ID = 'paper-restoration-v1-dream-pos-token-class-linear-probes'
RESULTS_ROOT = Path('/content/drive/MyDrive/dlmrel-paper-results/dream')
RUN_DIR = RESULTS_ROOT / 'exploratory_extensions/dream_7b/ewt/pos_token_class_linear_probes' / RUN_ID
CHECKOUT = Path('/content/latentrelationsondlm-dream-pos12')
LOCAL_CACHE = Path('/content/dlmrel-dream-pos-cache')
RUN_EXTRACTION = False  # Set True only in a GPU runtime when Dream extraction is missing.
TOTAL_BUDGET_SECONDS = 8 * 60 * 60
VALIDATION_RESERVE_SECONDS = 15 * 60

def run(*args, cwd=None):
    print('+', ' '.join(map(str, args)), flush=True)
    subprocess.run(list(map(str, args)), cwd=cwd, check=True, env=os.environ.copy())

drive.mount('/content/drive')
```

```python
if CHECKOUT.exists():
    shutil.rmtree(CHECKOUT)
run('git', 'clone', '--filter=blob:none', '--no-checkout', REPOSITORY, CHECKOUT)
run('git', 'fetch', 'origin', BRANCH, cwd=CHECKOUT)
run('git', 'checkout', '--detach', PROTOCOL_COMMIT, cwd=CHECKOUT)
head = subprocess.check_output(['git', 'rev-parse', 'HEAD'], cwd=CHECKOUT, text=True).strip()
assert head == PROTOCOL_COMMIT, (head, PROTOCOL_COMMIT)
run('python', '-m', 'pip', 'install', '-q', '-e', '.[dev]', cwd=CHECKOUT)
```

## Optional Dream GPU extraction
This is the only cell that may load Dream-7B. It is disabled by default. Existing valid extraction checkpoints are resumed. Fitting never invokes this cell.

```python
if RUN_EXTRACTION:
    import torch
    if not torch.cuda.is_available():
        raise RuntimeError('Dream extraction requires a GPU runtime')
    run('python', '-m', 'pip', 'install', '-q', '-r', 'requirements/dream.txt', cwd=CHECKOUT)
    command = ['python', '-m', 'dlmrel.cli', 'run', '--model', 'configs/models/dream_7b.yaml', '--dataset', 'configs/datasets/ewt.yaml', '--experiment', 'configs/experiments/pos_token_class_linear_probes.yaml', '--results', str(RESULTS_ROOT), '--run-id', RUN_ID, '--pos-stage', 'extract']
    if RUN_DIR.exists():
        command.append('--resume')
    run(*command, cwd=CHECKOUT)
    raise SystemExit('Dream extraction finished. Change to the 12-logical-CPU runtime, set RUN_EXTRACTION=False, and rerun the notebook.')
```

```python
os.environ.update({'CUDA_VISIBLE_DEVICES': '', 'OMP_NUM_THREADS': '1', 'OPENBLAS_NUM_THREADS': '1', 'MKL_NUM_THREADS': '1', 'NUMEXPR_NUM_THREADS': '1'})
if not RUN_DIR.is_dir():
    raise FileNotFoundError(f'Dream POS extraction is absent: {RUN_DIR}. Use the gated GPU extraction cell first.')
feature_parquets = list((RUN_DIR / 'checkpoints').glob('paper-pos-*-features*.parquet'))
if len(feature_parquets) != 180:
    raise RuntimeError(f'Expected 180 Dream feature chunks; found {len(feature_parquets)}. Resume extraction before fitting.')
logical_cpus = os.cpu_count() or 1
worker_counts = [n for n in (6, 8, 10, 12) if n <= logical_cpus] or [max(1, logical_cpus)]
print(json.dumps({'run_dir': str(RUN_DIR), 'feature_parquets': len(feature_parquets), 'logical_cpus': logical_cpus, 'workers_to_benchmark': worker_counts, 'free_gib': round(shutil.disk_usage('/content').free / 2**30, 2)}, indent=2))
run('python', '-m', 'pytest', '-q', 'tests/test_paper_pos_adaptive_12.py', cwd=CHECKOUT)
```

```python
run('python', '-m', 'dlmrel.cli', 'pos-fit-adaptive-12', '--run-dir', RUN_DIR, '--local-cache', LOCAL_CACHE, '--budget-seconds', str(TOTAL_BUDGET_SECONDS), '--validation-reserve-seconds', str(VALIDATION_RESERVE_SECONDS), '--worker-counts', *map(str, worker_counts), cwd=CHECKOUT)
```

```python
run('python', '-m', 'dlmrel.cli', 'validate-pos-adaptive-12', '--run-dir', RUN_DIR, cwd=CHECKOUT)
OUTPUT = RUN_DIR / 'pos_adaptive_12'
manifest = json.loads((OUTPUT / 'adaptive_manifest.json').read_text())
assert manifest['status'] == 'confirmed'
print(json.dumps({'model': 'Dream-7B', 'status': manifest['status'], 'elapsed_hours': manifest['elapsed_seconds'] / 3600, 'workers': manifest['workers'], 'choices': str(OUTPUT / 'pos_head_choices.csv'), 'trajectories': str(OUTPUT / 'selected_head_progress_trajectories.csv')}, indent=2))
```
